# Importing Essential Libraries

```python
import pandas as pd
```

# Reading CSV File

```python
df=pd.read_csv('INDIAN BANK-CLEANUP FILE.csv')
```

## Data Cleaning


```python
# Drop the last two columns
df.drop(df.columns[-2:], axis=1, inplace=True)

# Display the updated DataFrame
df
```

## Parsing Date

```python
df['DATE'] = pd.to_datetime(df['DATE'], format="%d-%m-%Y", errors='coerce')
```

```python
df.tail(10)
```

```python
df.dtypes
```

# Converting to numeric values

```python
df['DEBIT'] = pd.to_numeric(df['DEBIT'], errors='coerce')
df['CREDIT'] = pd.to_numeric(df['CREDIT'], errors='coerce')
df['BALANCE'] = pd.to_numeric(df['BALANCE'], errors='coerce')

df.tail()
```

```python
df.dtypes
```

## Handling NaN

```python
df['DEBIT'] = df['DEBIT'].fillna(0)
df['CREDIT'] = df['CREDIT'].fillna(0)
df['BALANCE'] = df['BALANCE'].fillna(0)

df.tail(5)
```

# Analyze Data

## Calculate Total Income

```python
Sum_credit = df['CREDIT'].sum()

print("Total Income:", Sum_credit)
```

## Calculate Total Expenses

```python
Sum_debit = df['DEBIT'].sum()

print("Total Expenses:", Sum_debit)
```

## Identifying Frequent Transactions

```python
top_frequent_transactions = df.groupby('DESCRIPTION').size().sort_values(ascending=False).head(5)

print("Top 5 Most Frequent Transaction Descriptions:")
print(top_frequent_transactions)
```

# Report Generation

```python
report_df = df[["DATE", "DEBIT", "CREDIT"]]
```

```python
report_df.dtypes
```

```python
report_df["DATE"] = report_df["DATE"].dt.strftime("%Y-%m")
```

```python
report_df
```

```python
report_df = report_df.groupby("DATE").sum().reset_index()
```

```python
report_df.columns
```

```python
report_df
```

```python
balance_df = df[["DATE", "DEBIT", "CREDIT", "BALANCE"]]
```

```python
balance_df
```

```python
balance_df["DATE"] = balance_df["DATE"].dt.strftime("%Y-%m")
```

```python
balance_df
```

```python
## Finding last transaction balance of each month

final_balance = balance_df.groupby("DATE")["BALANCE"].last()
```

```python
## merging final balance of each month with new dataframe

report_df["BALANCE"] = report_df["DATE"].map(final_balance)
```

```python
report_df
```

```python
## writing this dataframe to new csv file

report_df.to_csv('monthly_summary_INDIAN_BANK.csv', index=False)
```

## Identifying Inconsistencies in BALANCE

```python
balance = df.iloc[0]["BALANCE"]
credit = df.iloc[0]["CREDIT"]
debit = df.iloc[0]["DEBIT"]
initial_balance = balance - credit + debit
```

```python
list_balance = []
current_balance = initial_balance

for index, row in report_df.iterrows():
    total_balance = current_balance + row["CREDIT"] - row["DEBIT"]
    list_balance.append(total_balance)
    current_balance = total_balance 
```

```python
list_balance
```

```python
report_df["Calculated_Balance"] = list_balance
report_df
```

## Bonus Question(Finding inconsistency in balance)

```python
# function to check for inconsistency
def check_inconsistency(row, previous_balance):
    i_balance = previous_balance - row['DEBIT'] + row['CREDIT']
    return abs(row['BALANCE'] - i_balance) <= 0.01


# Adding a column for inconsistency
previous_balance = initial_balance
inconsistencies = []

for _, row in df.iterrows():
    is_consistent = check_inconsistency(row, previous_balance)
    inconsistencies.append(is_consistent)
    previous_balance = row['BALANCE']  # Update the previous balance for the next row

df['Consistency'] = inconsistencies

df
```

```python
# printing the rows where balance is inconsistent
df[df['Consistency'] == False]
```
